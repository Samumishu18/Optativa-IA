# Memoria de la práctica 1 – Teachable Machine

En esta práctica he creado un modelo que distingue imágenes de **ratones** y **teclados** con Teachable Machine. Después lo he usado en una página web para comprobar si reconocía bien las imágenes.


## 1. Crear y entrenar el modelo

Primero entré en [Teachable Machine](https://teachablemachine.withgoogle.com/) y elegí un proyecto de imágenes. Creé dos categorías: **Raton** y **Teclado**.

Para entrenarlo añadí **60 imágenes de ratones** y **42 imágenes de teclados**. En la configuración del entrenamiento aparecen **50 épocas**, un **tamaño de lote de 16** y una **tasa de aprendizaje de 0,001**. Después preparé el modelo con las imágenes que había añadido.

En esta captura se ven las dos categorías, las imágenes utilizadas y la configuración del entrenamiento.


Proyecto en Teachable Machine: Raton con 60 imágenes y Teclado con 42; configuración del entrenamiento


## 2. Exportar el modelo

Cuando terminé, usé **Exportar modelo** para descargarlo. Guardé una versión para **TensorFlow.js**, que es la que he utilizado en la página web, y otra para **Keras/TensorFlow**, por si la necesitamos en otra práctica.

En la versión para la web venían los archivos `model.json`, `metadata.json` y `weights.bin`. Los puse dentro de la carpeta `modelo` del proyecto.


## 3. Preparar la página web

Hice una página llamada **Vision Lab** con HTML, CSS y JavaScript y conecté el modelo de Teachable Machine.

Además del diseño, añadí las mejoras pedidas en el ejercicio: que se adapte al tamaño de la pantalla, botones para iniciar y detener la cámara, subida de imágenes, resultado con su porcentaje, un umbral de coincidencia y un historial de predicciones.

Para las pruebas que aparecen aquí utilicé la opción **Subir imagen**.


## 4. Abrir la web en localhost

Para ejecutar la página en local, **abrí CMD de Windows** y, desde la carpeta de la página web, ejecuté este comando:

```cmd
python -m http.server 8080
```

Después abrí **`http://localhost:8080`** en el navegador. La página cargó correctamente y apareció el aviso **Modelo preparado** con dos categorías disponibles.
